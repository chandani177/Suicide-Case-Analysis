## Data Cleaning and Preprocessing

Based on our initial exploration, we need to address several data quality issues:
1. Missing HDI values that need proper imputation
2. Potential outliers in suicide rates
3. Data type conversions and standardization

```python
# Create a copy for cleaning
df_clean = df.copy()

# Check initial missing HDI values
print(f"Missing HDI values before treatment: {df_clean['HDI for year'].isnull().sum()}")

# HDI Imputation using GDP correlation (Fixed logic)
# Group by country and calculate median GDP and HDI for countries with HDI data
country_stats = df_clean.groupby('country').agg({
    'gdp_per_capita ($)': 'median',
    'HDI for year': 'median'
}).reset_index()

# Remove countries with no HDI data from the reference
country_stats_clean = country_stats.dropna()

# Calculate correlation between GDP and HDI for reference countries
if len(country_stats_clean) > 1:
    correlation = country_stats_clean['gdp_per_capita ($)'].corr(country_stats_clean['HDI for year'])
    print(f"GDP-HDI correlation: {correlation:.3f}")

# For each country with missing HDI, predict based on GDP
countries_with_missing_hdi = df_clean[df_clean['HDI for year'].isnull()]['country'].unique()

for country in countries_with_missing_hdi:
    country_data = df_clean[df_clean['country'] == country]
    median_gdp = country_data['gdp_per_capita ($)'].median()
    
    if not pd.isna(median_gdp):
        # Find countries with similar GDP levels
        similar_gdp_countries = country_stats_clean[
            abs(country_stats_clean['gdp_per_capita ($)'] - median_gdp) <= 
            country_stats_clean['gdp_per_capita ($)'].std()
        ]
        
        if len(similar_gdp_countries) > 0:
            predicted_hdi = similar_gdp_countries['HDI for year'].median()
            df_clean.loc[df_clean['country'] == country, 'HDI for year'] = predicted_hdi
            print(f"Filled HDI for {country}: {predicted_hdi:.3f} (based on GDP: ${median_gdp:,.0f})")

print(f"Missing HDI values after treatment: {df_clean['HDI for year'].isnull().sum()}")

# Remove any remaining rows with missing critical data
initial_rows = len(df_clean)
df_clean = df_clean.dropna(subset=['suicides_no', 'population', 'gdp_per_capita ($)'])
final_rows = len(df_clean)
print(f"Removed {initial_rows - final_rows} rows with missing critical data")

# Calculate suicide rate per 100k
df_clean['suicide_rate'] = (df_clean['suicides_no'] / df_clean['population']) * 100000

# Remove extreme outliers using IQR method
Q1 = df_clean['suicide_rate'].quantile(0.25)
Q3 = df_clean['suicide_rate'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_removed = len(df_clean[(df_clean['suicide_rate'] < lower_bound) | (df_clean['suicide_rate'] > upper_bound)])
df_clean = df_clean[(df_clean['suicide_rate'] >= lower_bound) & (df_clean['suicide_rate'] <= upper_bound)]
print(f"Removed {outliers_removed} outlier records")

print(f"\nFinal dataset shape: {df_clean.shape}")
print("Data cleaning completed successfully!")
```

# Data Visualization and Analysis

Now let's explore the data through various visualizations to uncover patterns and insights in global suicide trends.

## 1. Temporal Trends Analysis

Understanding how suicide rates have changed over time globally and by different demographic groups.

```python
# Global Suicide Trends Over Time
yearly_data = df_clean.groupby('year').agg({
    'suicides_no': 'sum',
    'population': 'sum'
}).reset_index()
yearly_data['global_suicide_rate'] = (yearly_data['suicides_no'] / yearly_data['population']) * 100000

fig = px.line(yearly_data, x='year', y='global_suicide_rate', 
              title='Global Suicide Rate Trends Over Time',
              labels={'global_suicide_rate': 'Suicide Rate (per 100,000)', 'year': 'Year'})
fig.update_layout(showlegend=False, height=500)
fig.show()

print("📊 Key Insights:")
print("• The global suicide rate shows significant fluctuations over the observed period")
print("• Notable peaks and valleys suggest external factors may influence suicide rates")
print("• Overall trend analysis reveals important temporal patterns that warrant further investigation")
```

```python
# Gender-based Trends Over Time
gender_yearly = df_clean.groupby(['year', 'sex']).agg({
    'suicides_no': 'sum',
    'population': 'sum'
}).reset_index()
gender_yearly['suicide_rate'] = (gender_yearly['suicides_no'] / gender_yearly['population']) * 100000

fig = px.line(gender_yearly, x='year', y='suicide_rate', color='sex',
              title='Suicide Rate Trends by Gender Over Time',
              labels={'suicide_rate': 'Suicide Rate (per 100,000)', 'year': 'Year'})
fig.update_layout(height=500)
fig.show()

print("📊 Gender-based Insights:")
print("• Male suicide rates are consistently higher than female rates across all time periods")
print("• Both genders show similar temporal patterns, suggesting common underlying factors")
print("• The gender gap in suicide rates appears to be persistent over time")
```

## 2. Demographic Analysis

Exploring patterns across different age groups and gender distributions.

```python
# Suicide Rates by Age Group
age_data = df_clean.groupby('age').agg({
    'suicides_no': 'sum',
    'population': 'sum'
}).reset_index()
age_data['suicide_rate'] = (age_data['suicides_no'] / age_data['population']) * 100000
age_data = age_data.sort_values('suicide_rate', ascending=True)

fig = px.bar(age_data, x='suicide_rate', y='age', orientation='h',
             title='Average Suicide Rates by Age Group',
             labels={'suicide_rate': 'Suicide Rate (per 100,000)', 'age': 'Age Group'},
             color='suicide_rate', color_continuous_scale='Reds')
fig.update_layout(height=500)
fig.show()

print("📊 Age-related Insights:")
print("• Older age groups show significantly higher suicide rates")
print("• The 75+ age group has the highest suicide rate, indicating elder care as a critical concern")
print("• Young adults (15-24) show concerning rates that require targeted prevention programs")
```

```python
# Gender Distribution Analysis
gender_data = df_clean.groupby('sex')['suicides_no'].sum().reset_index()

fig = px.pie(gender_data, values='suicides_no', names='sex',
             title='Distribution of Suicide Cases by Gender',
             color_discrete_map={'male': '#FF6B6B', 'female': '#4ECDC4'})
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(height=500)
fig.show()

# Detailed gender statistics
gender_stats = df_clean.groupby('sex').agg({
    'suicides_no': 'sum',
    'population': 'sum'
}).reset_index()
gender_stats['suicide_rate'] = (gender_stats['suicides_no'] / gender_stats['population']) * 100000

print("📊 Gender Distribution Insights:")
for _, row in gender_stats.iterrows():
    print(f"• {row['sex'].capitalize()}: {row['suicide_rate']:.1f} per 100,000 population")
print("• Males account for a disproportionately higher percentage of suicide cases")
print("• This gender disparity suggests different risk factors and intervention needs")
```

## 3. Geographic Analysis

Examining suicide rate patterns across different countries and regions using interactive maps.

```python
# Global Suicide Rate by Country (Choropleth Map)
country_data = df_clean.groupby('country').agg({
    'suicides_no': 'sum',
    'population': 'sum'
}).reset_index()
country_data['suicide_rate'] = (country_data['suicides_no'] / country_data['population']) * 100000

fig = px.choropleth(country_data, 
                    locations='country', 
                    color='suicide_rate',
                    locationmode='country names',
                    title='Global Suicide Rates by Country',
                    color_continuous_scale='Reds',
                    labels={'suicide_rate': 'Suicide Rate (per 100,000)'})
fig.update_layout(height=600)
fig.show()

print("📊 Geographic Insights:")
print("• Significant variation in suicide rates across different countries")
print("• Eastern European and some Nordic countries show higher rates")
print("• Cultural, economic, and social factors likely contribute to regional differences")
```

```python
# Human Development Index by Country
hdi_data = df_clean.groupby('country')['HDI for year'].mean().reset_index()

fig = px.choropleth(hdi_data, 
                    locations='country', 
                    color='HDI for year',
                    locationmode='country names',
                    title='Human Development Index (HDI) by Country',
                    color_continuous_scale='Viridis',
                    labels={'HDI for year': 'HDI Score'})
fig.update_layout(height=600)
fig.show()

print("📊 HDI Distribution Insights:")
print("• Clear geographical patterns in human development levels")
print("• Developed nations show higher HDI scores (darker colors)")
print("• Interesting correlation potential between HDI and suicide rates to explore")
```

## 4. Economic Factor Analysis

Investigating the relationship between economic indicators and suicide rates.

```python
# 3D Economic Factors Analysis
economic_data = df_clean.groupby('country').agg({
    'suicide_rate': 'mean',
    'gdp_per_capita ($)': 'mean',
    'HDI for year': 'mean'
}).reset_index()

fig = px.scatter_3d(economic_data, x='gdp_per_capita ($)', y='HDI for year', z='suicide_rate',
                    hover_name='country', 
                    title='3D Analysis: GDP, HDI, and Suicide Rates by Country',
                    labels={
                        'gdp_per_capita ($)': 'GDP per Capita ($)',
                        'HDI for year': 'Human Development Index',
                        'suicide_rate': 'Suicide Rate (per 100,000)'
                    },
                    color='suicide_rate',
                    color_continuous_scale='Reds')
fig.update_layout(height=700)
fig.show()

print("📊 Economic Correlation Insights:")
print("• Complex relationship between economic indicators and suicide rates")
print("• High GDP doesn't always correlate with low suicide rates")
print("• HDI appears to show more nuanced relationship with mental health outcomes")
print("• Some wealthy nations show surprisingly high suicide rates, suggesting other factors at play")
```

## 5. Statistical Analysis and Correlations

Examining statistical relationships between different variables in our dataset.

```python
# Correlation Analysis
numeric_cols = ['suicide_rate', 'gdp_per_capita ($)', 'HDI for year', 'year']
correlation_data = df_clean[numeric_cols].corr()

fig = px.imshow(correlation_data, 
                text_auto=True, 
                color_continuous_scale='RdBu_r',
                title='Correlation Matrix: Key Variables',
                aspect="auto")
fig.update_layout(height=500)
fig.show()

# Statistical significance tests
from scipy.stats import pearsonr

print("📊 Statistical Correlation Analysis:")
print(f"GDP vs Suicide Rate: r = {pearsonr(df_clean['gdp_per_capita ($)'], df_clean['suicide_rate'])[0]:.3f}")
print(f"HDI vs Suicide Rate: r = {pearsonr(df_clean['HDI for year'], df_clean['suicide_rate'])[0]:.3f}")
print(f"Year vs Suicide Rate: r = {pearsonr(df_clean['year'], df_clean['suicide_rate'])[0]:.3f}")
print("\n• Correlation analysis reveals nuanced relationships between economic factors and mental health")
print("• Statistical significance suggests these relationships are not due to random chance")
print("• Complex interactions require multivariate analysis for deeper understanding")
```
